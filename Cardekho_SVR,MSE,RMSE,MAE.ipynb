```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler,LabelEncoder,MinMaxScaler
from sklearn.svm import SVR #FOR REGRESSION
from sklearn.metrics import recall_score,precision_score,accuracy_score,r2_score,f1_score,mean_absolute_error,root_mean_squared_error,mean_squared_error
```

```python
df=pd.read_csv(r'C:\Users\amank\Downloads\Learning_day_1\cardekho.csv')
df.head()
```

```python
df.isnull().sum()
```

```python
df.rename(columns={'mileage(km/ltr/kg)':'mileage'},inplace=True)
```

```python
df.drop_duplicates()
```

```python
df.drop(columns=['max_power','name'],inplace=True)
```

```python
df.head()
```

```python
label=LabelEncoder()
```

```python
df['encoded_fuel']=label.fit_transform(df['fuel'])
df['encoded_seller_type']=label.fit_transform(df['seller_type'])
df['encoded_transmission']=label.fit_transform(df['transmission'])
df['encoded_owner']=label.fit_transform(df['owner'])
```

```python
df.head()
```

```python
df.drop(columns=['fuel','seller_type','transmission','owner'],inplace=True)
```

```python
df.head()
```

```python
x=df[['year','selling_price','km_driven','mileage','engine','seats','encoded_fuel','encoded_seller_type','encoded_transmission','encoded_owner']]
y=df['selling_price']
```

```python
x=x.fillna(x.mean())

# YOU CAN USE THIS TOO  (x or y bouth filling null values nah values filling)

# y=y.fillna(y.median())
# y=y.fillna(y.mode()[0])
```

```python
df.head()
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
model=SVR(kernel='linear',C=0.5)
```

```python
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)
r2_score(y_test,y_pred)

mean_squared_error(y_test,y_pred)
root_mean_squared_error(y_test,y_pred)
mean_absolute_error(y_test,y_pred)
```
